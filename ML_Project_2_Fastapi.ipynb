# Step 4 : Create Pydantic model

```python
from pydantic import BaseModel

class StudentInput(BaseModel):

    Age: int
    Study_Hours: float
    Stress_Level: str
    Avg_Daily_Usage_Hours: float
    Daily_Unlocks: int
    Physical_Activity_Hours: float
    Sleep_Hours_Per_Night: float
    Gender: str
    Academic_Level: str
    Most_Used_Platform: str
    Purpose_Of_Use: str
    Grouped_country: str
```

# Step 5 : Create Fast API application

```python
app = FastAPI()
@app.get("/")
def home():
  return{"message":"Student Menatl Health API is running"}
```

# Step 6 : Create Prediction endpoint

```python
import pandas as pd

@app.post("/predict")
def predict_student(student: StudentInput):

    input_data = pd.DataFrame([student.model_dump()])

    prediction = model.predict(input_data)

    return {
        "Predicted_Mental_Health_Score": float(prediction[0])
    }
```

# Step 7 : Run the API server

```python
def run_server():
  uvicorn.run(app,host="0.0.0.0",port=8000)

thread = threading.Thread(target = run_server,daemon=True)
thread.start()
```
